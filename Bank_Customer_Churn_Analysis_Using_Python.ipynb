# Stage 2: Data Cleaning, Data Transformation and Feature Engineering

This stage focuses on improving data quality by handling duplicate records, treating missing values, standardizing column names, checking data types, identifying potential outliers, and performing data transformation and feature engineering. The cleaned dataset will then be validated through final exploratory data analysis (EDA).

```python
# Remove Duplicate Records

print("Dataset shape before removing duplicates:", df.shape)

df = df.drop_duplicates().copy()

print("Dataset shape after removing duplicates:", df.shape)
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
# Check Missing Values

print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Handle Missing Values

numeric_cols = df.select_dtypes(include="number").columns
categorical_cols = df.select_dtypes(include=["object", "category"]).columns

for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

for col in categorical_cols:
    if df[col].isnull().any():
        mode_value = df[col].mode()
        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value[0])

print("Missing value handling completed.")
```

```python
# Recheck the Missing Values

print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Data Transformation or Data Wrangling

# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Rename columns for consistency
df = df.rename(columns={
    "customerid": "customer_id",
    "creditscore": "credit_score",
    "numofproducts": "num_of_products",
    "hascrcard": "has_credit_card",
    "isactivemember": "is_active_member",
    "estimatedsalary": "estimated_salary"
})

# Display updated column names
print("Updated Column Names:")
print(df.columns.tolist())

# Display the first five rows
display(df.head())
```

```python
# Check Data Types

df.info()

print("\nData Types of Each Column:")
print(df.dtypes)
```

```python
# Feature Engineering: Create Churn Status

df["churn_status"] = df["exited"].map({
    0: "Stayed",
    1: "Exited"
})

print(df[["exited", "churn_status"]].head())
```

```python
# Detect Potential Outliers Using IQR

Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["age"] < lower_bound) |
    (df["age"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Potential Outliers:", len(outliers))
```

### Interpretation: Skewness, Box Plot and Outlier Detection

**Skewness:** The age distribution has a skewness value of 1.011, indicating a positively skewed distribution with a longer tail toward older customers.

**Box Plot:** The box plot shows that most customer ages are concentrated between approximately 32 and 44 years, with a median age of 37 years. Several data points appear beyond the upper whisker, indicating potential outliers among older customers.

**Outlier Detection:** Using the IQR method, Q1 is 32, Q3 is 44, and the IQR is 12. The lower bound is 14 years, and the upper bound is 62 years. A total of 359 customers have ages above the upper bound and are identified as potential outliers.

**Business Insight:** The analysis indicates that most customers are relatively young, while a smaller group of older customers extends the age distribution toward the right. These older customers may represent a distinct customer segment worth examining when investigating churn patterns.

**Outlier Treatment:** The potential outliers have not been removed because these ages may represent valid customers. Retaining them preserves potentially valuable information for further customer churn analysis.

Visual 1 — Churn Rate by Age Group and Geography
Business question: How does customer churn vary across age groups and countries?

```python
# Multivariate Visual 1: Churn Rate by Age Group and Geography

# Create age groups
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 30, 40, 50, 60, 120],
    labels=["30 and below", "31–40", "41–50", "51–60", "Above 60"]
)

# Calculate churn rate by age group and geography
churn_age_geo = df.pivot_table(
    index="Age_Group",
    columns="Geography",
    values="Exited",
    aggfunc="mean",
    observed=False
) * 100

# Plot heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    churn_age_geo,
    annot=True,
    fmt=".1f",
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Customer Churn Rate by Age Group and Geography")
plt.xlabel("Geography")
plt.ylabel("Age Group")

plt.show()

print("Churn rate by age group and geography (%):")
display(churn_age_geo.round(2))
```

Darker shades indicate higher churn rates, while lighter shades indicate lower churn rates.

**Features/Columns Used:**
- `Age`: Used to categorize customers into age groups.
- `Geography`: Represents the customer's country (France, Germany, or Spain).
- `Exited`: Indicates customer churn status, where 1 represents exited and 0 represents stayed.

**What does the chart show?**

The highest churn rate is observed among customers aged **51–60 in Germany, at 69.55%**, followed by customers aged 51–60 in France at 52.59% and Germany's 41–50 age group at 48.82%. Visual 2 — Churn Rate by Geography and Active Membership

```python
# Multivariate Visual 2: Churn Rate by Geography and Active Membership

df["Membership_Status"] = df["IsActiveMember"].map({
    0: "Inactive",
    1: "Active"
})

churn_geo_activity = (
    df.groupby(
        ["Geography", "Membership_Status"],
        observed=False
    )["Exited"]
    .mean()
    .reset_index()
)

churn_geo_activity["Churn_Rate"] = (
    churn_geo_activity["Exited"] * 100
)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=churn_geo_activity,
    x="Geography",
    y="Churn_Rate",
    hue="Membership_Status"
)

plt.title("Customer Churn Rate by Geography and Active Membership")
plt.xlabel("Geography")
plt.ylabel("Churn Rate (%)")
plt.legend(title="Membership Status")

plt.show()

print("Churn rates by geography and membership status (%):")
display(
    churn_geo_activity[
        ["Geography", "Membership_Status", "Churn_Rate"]
    ].round(2)
)
```

### Interpretation of Multivariate Visual 2: Customer Churn Rate by Geography and Active Membership
